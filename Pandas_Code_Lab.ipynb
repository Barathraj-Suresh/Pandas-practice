## 2. Series

A **Series** is a one-dimensional labelled data structure.

### Example

```python
marks = pd.Series([80, 75, 90, 65, 88])
print(marks)
```

### 🧪 Practice 1

```python
import pandas as pd

marks = pd.Series([85, 90, 78, 92, 88])

print(marks)
print("First mark:", marks[0])
print("Last mark:", marks[4])
print("Average:", marks.mean())
print("Maximum:", marks.max())
print("Minimum:", marks.min())
```

## 9. Missing Values

Create a DataFrame containing a missing mark.

```python
data_missing = {
    "Name": ["Arun", "Priya", "Rahul", "Divya"],
    "Marks": [85, None, 76, 90]
}

df_missing = pd.DataFrame(data_missing)
df_missing
```

Useful methods:

- `isnull()` → find missing values
- `isnull().sum()` → count missing values
- `fillna()` → replace missing values

```python
print(df_missing.isnull())
print(df_missing.isnull().sum())
```
