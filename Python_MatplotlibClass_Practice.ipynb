## 1. What is Matplotlib?

Matplotlib is a Python library used to create graphs and visualizations.

Think of it like this:

- **Python list / NumPy array** → contains the data
- **Matplotlib** → turns the data into a visual graph
- **Graph** → helps us understand patterns and comparisons

A common import is:

```python
import matplotlib.pyplot as plt
```

`pyplot` provides the plotting functions, and `plt` is the common short name.

**Teaching question:** Why do we visualize data instead of looking only at numbers?



````python
import matplotlib.pyplot as plt
print("Matplotlib imported successfully!")
````

## 2. First Plot — `plt.plot()`

The basic plotting function is:

```python
plt.plot(y)
```

If only Y-values are supplied, Matplotlib automatically uses X-values such as `0, 1, 2, 3...`.

Example:

```text
Y values: 10 20 15 30
X values:  0  1  2  3
```

So the graph connects the points:

`(0,10) → (1,20) → (2,15) → (3,30)`

**Key idea:** `plot()` connects data points with a line.



````python
import matplotlib.pyplot as plt

y = [10, 20, 15, 30]

plt.plot(y)
plt.show()
````

## 3. X-values and Y-values

You can explicitly provide both axes:

```python
plt.plot(x, y)
```

Example:

- X = `[1, 2, 3, 4]`
- Y = `[10, 20, 15, 30]`

Each X-value is paired with the Y-value at the same position.

So:
- `(1,10)`
- `(2,20)`
- `(3,15)`
- `(4,30)`

**Important:** X and Y should normally contain the same number of values.


````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 30]

plt.plot(x, y)
plt.show()
````

## 4. Markers

A marker highlights individual data points.

Example:

```python
plt.plot(y, marker='o')
```

Common markers include:
- `o` → circle
- `*` → star
- `s` → square
- `D` → diamond
- `^` → triangle

You can also change marker size using `ms`.

Example:

```python
plt.plot(y, marker='o', ms=10)
```

**Teaching idea:** A line shows the trend; markers make the individual observations easier to see.



````python
import matplotlib.pyplot as plt

y = [3, 8, 1, 10]

plt.plot(y, marker='d', ms=15)
plt.show()
````

## 5. Line Style, Color and Width

Matplotlib allows you to control the appearance of a line.

Important arguments:

```python
linestyle='dashed'
color='green'
linewidth=3
```

Short forms include:
- `ls` → linestyle
- `c` → color
- `lw` → linewidth

Common line styles:
- `'-'` → solid
- `':'` → dotted
- `'--'` → dashed
- `'-.'` → dash-dot



````python
import matplotlib.pyplot as plt

y = [3, 8, 1, 10]

plt.plot(y, linestyle='dashed', color='green', linewidth=2)
plt.show()
````

## 6. Titles and Axis Labels

A graph should communicate what the data means.

Use:

```python
plt.title("Student Marks")
plt.xlabel("Subject")
plt.ylabel("Marks")
```

This makes the graph understandable without explaining it verbally.

**Rule:** A good graph should answer:
- What is this graph?
- What does the X-axis represent?
- What does the Y-axis represent?


````python
import matplotlib.pyplot as plt

subjects = [1, 2, 3, 4, 5]
marks = [78, 85, 72, 90, 88]

plt.plot(subjects, marks, marker='o',c='r')
plt.title("Student Marks")
plt.xlabel("Subject Number")
plt.ylabel("Marks")
plt.show()
````

## 7. Grid

A grid makes values easier to read.

Use:

```python
plt.grid()
```

You can also customize the grid, but first understand the basic function.

**Question:** When would a grid be useful?

Answer: When the reader needs to estimate values from the graph.


````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 25, 18, 30, 22]

plt.plot(x, y, marker='o')
plt.title("Sample Data")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid()
plt.show()
````

## 8. Multiple Lines

You can plot more than one dataset on the same graph.

Example:

```python
plt.plot(x, y1)
plt.plot(x, y2)
```

This is useful for comparisons.

Example: compare two students' marks or two products' sales.


````python
import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5]

student_a = [70, 75, 80, 78, 85]
student_b = [65, 72, 76, 82, 80]

plt.plot(months, student_a, marker='o', label='Student A')
plt.plot(months, student_b, marker='s', label='Student B')

plt.title("Student Performance")
plt.xlabel("Month")
plt.ylabel("Marks")
plt.legend()
plt.grid()
plt.show()
````

## 9. Subplots

A subplot lets you put multiple graphs inside one figure.

The basic syntax is:

```python
plt.subplot(rows, columns, position)
```

For example:

```python
plt.subplot(1, 2, 1)
```

means:
- 1 row
- 2 columns
- first graph

And:

```python
plt.subplot(1, 2, 2)
```

means the second graph.



````python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [10, 20, 15, 30]
y2 = [5, 15, 25, 20]

plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.title("Graph 1")

plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.title("Graph 2")

plt.show()
````

## 10. Scatter Plot

A scatter plot displays individual points rather than connecting them with a line.

Use:

```python
plt.scatter(x, y)
```

Scatter plots are useful for examining relationships between two variables.

Example:
- hours studied vs marks
- advertising cost vs sales
- height vs weight


````python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [45, 50, 58, 65, 72, 85]

plt.scatter(hours, marks)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid()
plt.show()
````

## 12. Histogram

A histogram shows the distribution of numerical data.

Use:

```python
plt.hist(data)
```

Unlike a bar chart, a histogram groups numerical values into ranges called **bins**.

Example: exam marks can be grouped into ranges such as:
- 0–20
- 20–40
- 40–60
- 60–80
- 80–100


````python
import matplotlib.pyplot as plt

marks = [45, 52, 56, 60, 61, 65, 68, 70, 72, 75, 78, 80, 82, 85, 88, 90, 92, 95]

plt.hist(marks, bins=5)
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
````

## 13. Pie Chart

A pie chart shows how a whole is divided into categories.

Use:

```python
plt.pie(values, labels=labels)
```

A pie chart is most useful when:
- categories are limited
- values represent parts of a whole
- the comparison does not require very precise values


````python
import matplotlib.pyplot as plt

labels = ["Python", "Java", "C", "Other"]
values = [40, 25, 20, 15]

plt.pie(values, labels=labels, autopct='%1.1f%%')
plt.title("Programming Language Preference")
plt.show()
````

````python
# MINI PROJECT — YOUR CODE
subjects = ["Python", "Maths", "DBMS", "AI", "English"]
marks = [85, 78, 90, 88, 75]

# Build your three visualizations here.
import matplotlib.pyplot as plt

subjects = ["Python", "Maths", "DBMS", "AI", "English"]
marks = [85, 78, 90, 88, 75]

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.plot(subjects, marks, marker="o")
plt.title("Marks Across Subjects")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid()

plt.subplot(1, 3, 2)
plt.bar(subjects, marks)
plt.title("Subject-wise Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.subplot(1, 3, 3)
plt.pie(marks, labels=subjects, autopct="%1.1f%%")
plt.title("Marks Proportion")

plt.tight_layout()
plt.show()
````
